```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def hanning_window(N):
    n = np.arange(N)
    return 0.5 * (1 - np.cos(2 * np.pi * n / (N - 1)))
```

```python
# Example usage:
N = 100
window = hanning_window(N)
plt.plot(window)
plt.title('Hanning Window')
plt.xlabel('Sample')
plt.ylabel('Amplitude')
plt.show()
```
